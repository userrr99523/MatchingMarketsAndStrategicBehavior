```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import json
from req_functions_from_ApproxSP import *
from tqdm import tqdm
from generate_profiles import *
from euclidean import *
from mallows import *
from impartial import *
from all_stable import *
import pandas as pd
import ast
import csv
```

**Euclidean Mallows**

```python
df = pd.read_csv('euclidean_mallows_accomplice.csv')

self_manipulation_count = []
    
for i, row in df.iterrows():
    mpref = json.loads(row['mpref'])
    wpref = json.loads(row['wpref'])

    n = len(mpref)
    count = 0
    original_matching = gale_shapley(mpref, wpref)
    for manipulator in range(1, n + 1):
        manipulated_matching = self_manipulation(mpref, wpref, manipulator)
        if manipulated_matching[0] != original_matching:
            count += 1
    self_manipulation_count.append(count)

df['count_self_manipulating_women'] = self_manipulation_count
df.to_csv('euclidean_mallows_accomplice.csv', index=False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

summary = df.groupby(['phi_m', 'phi_w']).agg({'count_self_manipulating_women': 'mean'}).reset_index()

pivot_table = summary.pivot(index = 'phi_m', columns = 'phi_w', values = 'count_self_manipulating_women')
sns.heatmap(pivot_table, annot=True, cmap="viridis")
plt.gca().invert_yaxis() 
plt.title("Average Number of Self Manipulating Women")
plt.xlabel("$\phi_w$")
plt.ylabel("$\phi_m$")
plt.savefig("self_manip_euclidean_mallows.pdf", format='pdf')
plt.show()
```
