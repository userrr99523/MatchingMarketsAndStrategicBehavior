```python
import pandas as pd

df1 = pd.read_csv("two_for_one_results_Euclidean")
df2 = pd.read_csv("backup_euclidean_mallows.csv")

# Merge by row index
df2["Two_for_one"] = df1["Two_for_one"]

df2.to_csv("backup_euclidean_mallows.csv", index=False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('backup_euclidean_mallows.csv')

summary = df.groupby(['phi_m', 'phi_w']).agg({'Two_for_one': 'mean'}).reset_index()

pivot_table = summary.pivot(index = 'phi_m', columns = 'phi_w', values = 'Two_for_one')
sns.heatmap(pivot_table, annot=True, cmap="viridis")
plt.gca().invert_yaxis() 
plt.title("Average Number of Manipulators")
plt.xlabel("$\phi_w$")
plt.ylabel("$\phi_m$")
#plt.savefig("one_for_all_euclidean_mallows.pdf", format='pdf')
plt.show()
```
