```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the CSV
df = pd.read_csv("mallows_MA_gain.csv")

# Create trunc_prevalence column: 1 if stability_count > 1, else 0
df["trunc_prevalence"] = (df["stability_count"] > 1).astype(int)

df.to_csv("mallows_MA_gain.csv", index=False)

# Compute average trunc_prevalence grouped by phi_m and phi_w
heatmap_data = df.groupby(["phi_m", "phi_w"])["trunc_prevalence"].mean().reset_index()

# Pivot for heatmap
pivot_table = heatmap_data.pivot(index="phi_m", columns="phi_w", values="trunc_prevalence")

# Plot heatmap
plt.figure(figsize=(8, 6))
ax = sns.heatmap(pivot_table, annot=True, fmt=".2f", cmap="viridis", annot_kws={"size": 22})
#plt.title("Average trunc_prevalence by phi_m and phi_w")
plt.xlabel("$\phi_w$", fontsize=30)
plt.ylabel("$\phi_m$", fontsize=30)
plt.gca().invert_yaxis()
plt.tight_layout()

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)

colorbar = ax.collections[0].colorbar
colorbar.ax.tick_params(labelsize=16)

# Save as PDF
plt.savefig("trunc_prevalence_mall_MA_new.pdf", format="pdf")

plt.show()
```
