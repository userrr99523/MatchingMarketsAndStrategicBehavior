```python
import pandas as pd
```

**Euclidean Mallows**

```python

df = pd.read_csv("two_for_one_results_euclidean_mallows")

# Step 1: Add a boolean column for manipulation
df['manipulated'] = (df['Two_for_one'] != 0).astype(int)

# Step 2: Group by phi_1 and phi_2, and calculate counts
grouped = df.groupby(['Phi1', 'Phi2'])

# Step 3: Compute total and number of manipulated cases
summary = grouped['manipulated'].agg(
    total='count',
    manipulated='sum'
).reset_index()

# Step 4: Compute percentage
summary['manipulation_percentage'] = 100 * summary['manipulated'] / summary['total']

```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Pivot for heatmap
heatmap_data = summary.pivot(index='Phi1', columns='Phi2', values='manipulation_percentage')

# Plot
plt.figure(figsize=(10, 8))
sns.heatmap(heatmap_data, annot=True, fmt=".1f", cmap="viridis")

# Invert y-axis (optional)
plt.gca().invert_yaxis()

# Labels
plt.title("Percentage of Two for One")
plt.xlabel(r'$\phi_w$')
plt.ylabel(r'$\phi_m$')
plt.tight_layout()
plt.show()
```
