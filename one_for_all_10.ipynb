**Mallows MA**

```python
df = pd.read_csv('mallows_MA_gain.csv')

count_men = []

for idx in range(len(df)):
    row = df.loc[idx]
    mpref, wpref = ast.literal_eval(row["preferences"]) 


    matching = ast.literal_eval(row["original matchings"])
    n = len(mpref)
    count = 0
    
    for accomplice in range (1, n+1):
        for manipulator in range(1, n+1):
            result = accomplice_manipulation(mpref, wpref, manipulator, accomplice)
            if result [0]!= matching[0]:
                count += 1
                break

    count_men.append(count)

df['count_men'] = count_men

df.to_csv('mallows_MA_gain.csv', index=False)           
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the CSV
df = pd.read_csv("mallows_MA_gain.csv")

# Compute average trunc_prevalence grouped by phi_m and phi_w
heatmap_data = df.groupby(["phi_m", "phi_w"])["count_men"].mean().reset_index()

# Pivot for heatmap
pivot_table = heatmap_data.pivot(index="phi_m", columns="phi_w", values="count_men")

# Plot heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(pivot_table, annot=True, fmt=".2f", cmap="viridis")
#plt.title("Average trunc_prevalence by phi_m and phi_w")
plt.xlabel("$\phi_w$", fontsize=20)
plt.ylabel("$\phi_m$", fontsize=20)
plt.gca().invert_yaxis()
plt.tight_layout()

# Save as PDF
#plt.savefig("trunc_prevalence_mall_MA_new.pdf", format="pdf")

plt.show()
```
